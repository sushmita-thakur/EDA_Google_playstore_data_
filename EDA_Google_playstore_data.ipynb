####  Start Operation on Size column

```python
df['Size'].unique()
```

```python
item="19M"
item[-1]
```

```python
# 1M =1024 kb 
def size_process(item):
    s = str(item).strip()
    if s.endswith('M'):
        try: return float(s[:-1]) * 1024
        except: return np.nan
    elif s.endswith('k'):
        try: return float(s[:-1])
        except: return np.nan
    return np.nan

# Comma pehle hatayein jab data string format me ho:
df_copy['Size'] = df_copy['Size'].astype(str).str.replace(',', '', regex=False)
df_copy['Size'] = df_copy['Size'].apply(size_process)
```
